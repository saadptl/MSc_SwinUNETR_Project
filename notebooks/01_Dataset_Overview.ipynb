```python
# ============================================================
# Dataset Path
# ============================================================

dataset_path = r"D:\Msc Major Project Swin Unetr Framework\MSc_SwinUNETR_Project\dataset\rsna-2024-lumbar-spine-degenerative-classification"

print("Dataset Path:")
print(dataset_path)
```

```python

# ============================================================
# Verify Dataset
# ============================================================

if os.path.exists(dataset_path):
    print("Dataset found successfully.")
else:
    print("Dataset path does not exist.")

    
```

```python
# ============================================================
# Display Folder Contents
# ============================================================

files = sorted(os.listdir(dataset_path))

print("Files and folders inside dataset:\n")

for item in files:
    print(item)
```

```python
# ============================================================
# Load CSV Files
# ============================================================

train_df = pd.read_csv(
    os.path.join(dataset_path, "train.csv")
)

series_df = pd.read_csv(
    os.path.join(dataset_path, "train_series_descriptions.csv")
)

coord_df = pd.read_csv(
    os.path.join(dataset_path, "train_label_coordinates.csv")
)
```

```python
# ============================================================
# Dataset Shapes
# ============================================================

print("Train CSV Shape:")
print(train_df.shape)

print("\nSeries CSV Shape:")
print(series_df.shape)

print("\nCoordinate CSV Shape:")
print(coord_df.shape)
```

```python
# ============================================================
# Display First Five Rows
# ============================================================

display(train_df.head())

display(series_df.head())

display(coord_df.head())
```

```python
# ============================================================
# Basic Statistics
# ============================================================

print("Number of Patients :", train_df.shape[0])

print("Number of Disease Columns :", train_df.shape[1]-1)

print("Number of MRI Series :", len(series_df))

print("Number of Coordinate Labels :", len(coord_df))
```
